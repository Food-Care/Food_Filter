```python
import json
import pandas as pd
import os

path = r"C:\Users\redmo\development\FOOD_Filter\dev\Data_Storage\정제 데이터\식품데이터_최종_표준화_완료.json"

print("현재 작업 디렉토리:", os.getcwd())
print("파일 존재 여부:", os.path.exists(path))
print("ABS_PATH:", path)

with open(path, "r", encoding="utf-8") as f:
    data = json.load(f)

df = pd.DataFrame(data)

company_names = df["회사명"].astype(str)
mask = company_names.str.contains(r"주식회사|\(주\)|㈜", regex=True)
filtered_companies = sorted(company_names[mask].unique())

print(f"📦 '주식회사' 표기 포함 회사 총 {len(filtered_companies)}개\n")
i=1
for name in filtered_companies:
    print(i, name)
    i=i+1
```

```python
# ✅ 1. 조건부 표기 통일 함수 정의
def normalize_company_name(name: str) -> str:
    name = str(name)
    if any(tag in name for tag in ["주식회사", "(주)", "㈜"]):
        # 주식회사, (주), ㈜ 포함된 경우만 처리
        clean_name = name.replace("주식회사", "").replace("(주)", "").replace("㈜", "").strip()
        return f"{clean_name}㈜"
    else:
        # 표기 없는 회사는 그대로 유지
        return name

# ✅ 2. 해당 규칙 적용
df["회사명"] = df["회사명"].apply(normalize_company_name)

# ✅ 3. 결과 예시 출력
print("✅ 표기 통일 완료. 예시 30개 출력:\n")
for i, name in enumerate(df["회사명"].unique(), start=1):
    print(i, name)

# ✅ 4. (선택) 저장
output_path = r"C:\Users\redmo\development\FOOD_Filter\dev\Data_Storage\정제 데이터\식품데이터_회사명_정제.json"
df.to_json(output_path, orient="records", force_ascii=False, indent=2)

print(f"\n💾 저장 완료: {output_path}")
```
