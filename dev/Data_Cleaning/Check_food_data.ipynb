```python
# 원재료명 확인 코드
import json

def find_ingredients_with_keywords(data_path, keywords_to_find):
    """
    JSON 데이터 파일에서 특정 키워드를 포함하는 모든 원재료명을 찾아 출력합니다.

    :param data_path: 원본 식품 데이터 JSON 파일 경로
    :param keywords_to_find: 검색할 키워드 리스트
    """
    try:
        with open(data_path, "r", encoding="utf-8") as f:
            food_data = json.load(f)
        print(f"✅ '{data_path}' 파일을 성공적으로 불러왔습니다.")
    except FileNotFoundError:
        print(f"🚨 파일 오류: '{data_path}' 파일을 찾을 수 없습니다. 파일명을 확인해주세요.")
        return

    found_ingredients = set()

    for product in food_data:
        ingredients = product.get("원재료명", [])
        if not ingredients:
            continue

        for ingredient in ingredients:
            for keyword in keywords_to_find:
                if keyword in ingredient:
                    found_ingredients.add(ingredient)
                    break

    if found_ingredients:
        print(f"\n--- 🔍 총 {len(found_ingredients)}개의 원재료명을 찾았습니다 ---")
        for item in sorted(list(found_ingredients)):
            print(item)
    else:
        print("\n--- 해당 키워드를 포함하는 원재료를 찾지 못했습니다. ---")


if __name__ == "__main__":
    SEARCH_KEYWORDS = ["복숭"]

    print(f"지정한 키워드: {SEARCH_KEYWORDS}")
    
    input_file = "../Data_Storage/정제 데이터/식품데이터_최종_표준화_완료.json"

    find_ingredients_with_keywords(input_file, SEARCH_KEYWORDS)
```

```python
import json

def load_json_data(file_path):
    """지정된 경로의 JSON 파일을 읽어 데이터를 반환합니다."""
    try:
        with open(file_path, 'r', encoding='utf-8') as f:
            return json.load(f)
    except FileNotFoundError:
        print(f"🚨 오류: '{file_path}' 파일을 찾을 수 없습니다.")
        return None
    except json.JSONDecodeError:
        print(f"🚨 오류: '{file_path}' 파일이 올바른 JSON 형식이 아닙니다.")
        return None

def get_unique_individual_ingredients(food_data):
    """전체 데이터에서 고유한 '개별 원재료'의 set을 반환합니다."""
    if not food_data: return set()
    all_ingredients = set()
    for item in food_data:
        if isinstance(item, dict) and '원재료명' in item and item['원재료명']:
            ingredient_data = item['원재료명']
            if isinstance(ingredient_data, list):
                ingredient_string = ', '.join(map(str, ingredient_data))
            else:
                ingredient_string = str(ingredient_data)
            individual_items = ingredient_string.split(',')
            for ingredient in individual_items:
                cleaned_ingredient = ingredient.strip()
                if cleaned_ingredient:
                    all_ingredients.add(cleaned_ingredient)
    return all_ingredients

def find_unmatched_ingredients(ingredients_set, allergy_db):
    """
    알레르기 규칙(키워드, 예외)에 해당하지 않는 원재료만 필터링하여 반환합니다.
    """
    # 키워드나 예외 키워드에 하나라도 걸리는 원재료를 저장할 set
    filtered_out_ingredients = set()

    for ingredient in ingredients_set:
        is_filtered = False
        for allergen, data in allergy_db.items():
            # 키워드 검사
            if "keyword" in data:
                for keyword in data["keyword"]:
                    if keyword in ingredient:
                        filtered_out_ingredients.add(ingredient)
                        is_filtered = True
                        break
            if is_filtered: break

            # 예외 키워드 검사 (키워드에 걸리지 않았을 경우에만)
            if "exclusions" in data:
                for exclusion in data["exclusions"]:
                    if exclusion in ingredient:
                        filtered_out_ingredients.add(ingredient)
                        is_filtered = True
                        break
            if is_filtered: break

    # 전체 집합에서 걸러진 것들을 제외하여 최종 목록 생성
    unmatched_set = ingredients_set.difference(filtered_out_ingredients)
    return unmatched_set


# --- 메인 실행 부분 ---
if __name__ == "__main__":
    # 1. 파일 경로 설정
    allergy_rules_path = '../Data_Storage/표준화 데이터/Allergy_ver_2.json'
    food_data_path = '../Data_Storage/정제 데이터/식품데이터_최종_표준화_완료.json'

    # 2. 데이터 로드
    allergy_data = load_json_data(allergy_rules_path)
    food_data_json = load_json_data(food_data_path)

    if allergy_data and food_data_json:
        # 3. 고유한 개별 원재료 목록 준비
        unique_ingredients = get_unique_individual_ingredients(food_data_json)
        print(f"✅ 분석 준비 완료: 총 {len(unique_ingredients)}개의 고유 개별 원재료를 찾았습니다.")

        # 4. 규칙에 걸러지지 않은 원재료 필터링
        unmatched_list = find_unmatched_ingredients(unique_ingredients, allergy_data)

        # 5. 최종 결과 출력
        print("\n" + "="*50)
        print(f"📊 최종 분석 결과")
        print("="*50)
        print(f"키워드 규칙에 걸러지지 않은 고유 원재료는 총 {len(unmatched_list)}개입니다.")
        print("아래는 해당 원재료의 전체 목록입니다.")
        print("-"*50 + "\n")

        # 정렬하여 전체 목록 출력
        for ingredient_item in sorted(list(unmatched_list)):
            print(ingredient_item)
```

```python
import json
import pandas as pd

def refine_ingredients(food_data_path, allergy_data_path, output_path):
    """
    식품 데이터의 원재료를 알레르기 정보에 따라 정제하여 CSV 파일로 저장합니다.

    Args:
        food_data_path (str): 식품 데이터 JSON 파일 경로
        allergy_data_path (str): 알레르기 정보 JSON 파일 경로
        output_path (str): 결과를 저장할 CSV 파일 경로
    """
    try:
        # 데이터 불러오기
        with open(food_data_path, 'r', encoding='utf-8') as f:
            food_data = json.load(f)

        with open(allergy_data_path, 'r', encoding='utf-8') as f:
            allergy_data = json.load(f)

        # 모든 원재료 추출 및 중복 제거
        all_ingredients = set()
        for item in food_data:
            for ingredient in item.get('원재료명', []):
                all_ingredients.add(ingredient)

        # 키워드와 알레르기 정보 매핑
        keyword_to_allergy = {}
        for allergy, data in allergy_data.items():
            for keyword in data.get('keyword', []):
                keyword_to_allergy[keyword] = allergy

        # 데이터 가공
        processed_data = []
        for ingredient in sorted(list(all_ingredients)):
            found_allergy = False
            for keyword, allergy in keyword_to_allergy.items():
                if keyword in ingredient:
                    # 제외 키워드 확인
                    is_excluded = False
                    if 'exclusions' in allergy_data[allergy]:
                        for exclusion in allergy_data[allergy]['exclusions']:
                            if exclusion in ingredient:
                                is_excluded = True
                                break
                    
                    if not is_excluded:
                        processed_data.append({
                            '원재료명': ingredient,
                            '수정 원재료명': allergy,
                            '키워드': keyword,
                            '비고': ''
                        })
                        found_allergy = True
                        break  # 다음 원재료로 이동

            if not found_allergy:
                processed_data.append({
                    '원재료명': ingredient,
                    '수정 원재료명': '',
                    '키워드': '',
                    '비고': ''
                })

        # 데이터프레임 생성 및 엑셀 파일로 저장
        df = pd.DataFrame(processed_data)
        df.to_excel(output_path, index=False)
        print(f"'{output_path}' 파일이 성공적으로 생성되었습니다.")

    except FileNotFoundError as e:
        print(f"오류: 파일을 찾을 수 없습니다. 경로를 확인해주세요: {e}")
    except Exception as e:
        print(f"오류가 발생했습니다: {e}")

# --- 코드 실행 ---
# 입력 파일 경로
food_data_file = '../Data_Storage/정제 데이터/식품데이터_최종_표준화_완료.json'
allergy_data_file = '../Data_Storage/표준화 데이터/Allergy_ver_2.json'

# 출력 파일 경로
output_xlsx_file = 'refined_ingredients_ver1.xlsx'

# 함수 호출
refine_ingredients(food_data_file, allergy_data_file, output_xlsx_file)
```

데이터 키워드 태깅(프로토 타입)

```python
import json
import pandas as pd
from tqdm import tqdm # 진행 상황을 시각적으로 보여주는 라이브러리

def classify_ingredient_advanced(ingredient_name, allergy_data):
    """
    주어진 원재료명을 진화된 규칙(가장 긴 단어 우선)에 따라 분류하여
    알레르기 키 목록을 반환합니다.
    """
    candidate_details = {}

    for allergy_key, rules in allergy_data.items():
        matched_keywords = set()
        matched_exclusions = set()
        is_candidate = False

        # 모든 종류의 키워드 탐색
        keyword_lists = {
            'only_one': rules.get('only_one', []),
            'first_word': rules.get('first_word', []),
            'last_word': rules.get('last_word', []),
            'keyword': rules.get('keyword', [])
        }

        if ingredient_name in keyword_lists['only_one']:
            matched_keywords.add(ingredient_name)
            is_candidate = True
        
        for kw in keyword_lists['first_word']:
            if ingredient_name.startswith(kw):
                matched_keywords.add(kw)
                is_candidate = True
        
        for kw in keyword_lists['last_word']:
            if ingredient_name.endswith(kw):
                matched_keywords.add(kw)
                is_candidate = True

        for kw in keyword_lists['keyword']:
            if kw in ingredient_name:
                matched_keywords.add(kw)
                is_candidate = True
        
        if is_candidate:
            for ex_word in rules.get('exclusions', []):
                if ex_word in ingredient_name:
                    matched_exclusions.add(ex_word)
            
            candidate_details[allergy_key] = {
                "keywords": matched_keywords,
                "exclusions": matched_exclusions
            }
            
    final_allergies = []
    
    for allergy_key, details in candidate_details.items():
        if not details['exclusions']:
            final_allergies.append(allergy_key)
            continue
            
        longest_keyword_len = max(len(kw) for kw in details['keywords']) if details['keywords'] else 0
        longest_exclusion_len = max(len(ex) for ex in details['exclusions']) if details['exclusions'] else 0
        
        if longest_keyword_len >= longest_exclusion_len:
            final_allergies.append(allergy_key)
            
    return final_allergies

# --- 코드 실행 부분 ---
try:
    # 1. 알레르기 규칙 파일 로드 (Allergy_ver_2.json)
    with open('../Data_Storage/표준화 데이터/Allergy_ver_2.json', 'r', encoding='utf-8') as f:
        allergy_data = json.load(f)
    print("✅ 'Allergy_ver_2.json' 규칙 파일을 성공적으로 불러왔습니다.")

    # 2. 전처리 완료된 식품 데이터 로드
    with open('식품데이터_최종_표준화_ver3.json', 'r', encoding='utf-8') as f:
        food_data = json.load(f)
    print(f"✅ 총 {len(food_data)}개의 제품 데이터를 불러왔습니다. 알레르기 분석을 시작합니다.")

    # 3. 각 제품에 알레르기 정보 태깅
    # tqdm을 사용하여 진행률 표시
    for product in tqdm(food_data, desc="알레르기 정보 분석 중"):
        product_allergies = set()
        if '원재료명' in product and isinstance(product['원재료명'], list):
            for ingredient in product['원재료명']:
                found_allergies = classify_ingredient_advanced(ingredient, allergy_data)
                for allergy in found_allergies:
                    product_allergies.add(allergy)
        
        # 제품 정보에 "알레르기" 키로 최종 결과 추가
        product['알레르기'] = sorted(list(product_allergies))

    # 4. 최종 결과를 새로운 JSON 파일로 저장
    output_filename = '식품데이터_알레르기_분석_완료.json'
    with open(output_filename, 'w', encoding='utf-8') as f:
        json.dump(food_data, f, ensure_ascii=False, indent=4)

    print(f"\n🎉 모든 작업 완료! 최종 결과가 '{output_filename}' 파일에 저장되었습니다.")

except FileNotFoundError as e:
    print(f"❌ 파일 오류: '{e.filename}' 파일을 찾을 수 없습니다. 필요한 파일이 모두 있는지 확인해주세요.")
except Exception as e:
    print(f"오류가 발생했습니다: {e}")
```

```python
import json

try:
    # --- 데이터 로드 ---
    # 1. 알레르기 규칙 파일 로드 (Allergy_ver_2.json)
    with open('../Data_Storage/표준화 데이터/Allergy_ver_2.json', 'r', encoding='utf-8') as f:
        allergy_data = json.load(f)
    print("✅ 'Allergy_ver_2.json' 규칙 파일을 불러왔습니다.")

    # 2. 분석할 식품 데이터 로드
    with open('식품데이터_최종_표준화_ver3.json', 'r', encoding='utf-8') as f:
        food_data = json.load(f)
    print(f"✅ 총 {len(food_data)}개의 제품 데이터를 불러왔습니다.\n")

    # --- 1. '소'로 시작하는 모든 원재료명 찾기 ---
    so_starting_ingredients = set()
    for product in food_data:
        if '원재료명' in product and isinstance(product['원재료명'], list):
            for ingredient in product['원재료명']:
                # '소'로 시작하고, 글자 길이가 1보다 큰 경우 (단일 '소'는 제외)
                if ingredient.startswith('소') and len(ingredient) > 1:
                    so_starting_ingredients.add(ingredient)

    print("="*50)
    print("### 1. '소'로 시작하는 모든 원재료명 목록 ###")
    print("="*50)
    if so_starting_ingredients:
        # 가나다 순으로 정렬하여 출력
        for ingredient in sorted(list(so_starting_ingredients)):
            print(ingredient)
    else:
        print("-> '소'로 시작하는 원재료명을 찾을 수 없습니다.")
    print("\n")


    # --- 2. '소' 시작 원재료가 포함된 모든 제품 출력 ---
    products_with_so_starting = []
    for product in food_data:
        found = False
        if '원재료명' in product and isinstance(product['원재료명'], list):
            for ingredient in product['원재료명']:
                if ingredient.startswith('소') and len(ingredient) > 1:
                    products_with_so_starting.append(product)
                    found = True
                    break # 제품 하나당 한 번만 추가
    
    print("="*50)
    print("### 2. '소' 시작 원재료가 포함된 모든 제품 목록 ###")
    print("="*50)
    if products_with_so_starting:
        # JSON 형식으로 보기 좋게 출력
        print(json.dumps(products_with_so_starting, ensure_ascii=False, indent=4))
    else:
        print("-> '소'로 시작하는 원재료가 포함된 제품을 찾을 수 없습니다.")
    print("\n")


    # --- 3. 단일 '소' 키워드로 걸러진 제품 출력 ---
    products_with_single_so = []
    # Allergy_ver_2.json에서 "쇠고기"의 "only_one" 규칙에 "소"가 있는지 확인
    so_only_one_rule_exists = '소' in allergy_data.get('쇠고기', {}).get('only_one', [])
    
    if so_only_one_rule_exists:
        for product in food_data:
            if '원재료명' in product and isinstance(product['원재료명'], list):
                # 원재료명 리스트에 '소'가 정확히 포함된 경우
                if '소' in product['원재료명']:
                    products_with_single_so.append(product)

    print("="*50)
    print("### 3. 단일 글자 '소'가 원재료명에 포함된 제품 목록 ###")
    print("='only_one' 규칙으로 인해 '쇠고기'로 분류될 가능성이 있는 제품들")
    print("="*50)
    if products_with_single_so:
        print(json.dumps(products_with_single_so, ensure_ascii=False, indent=4))
    else:
        print("-> 원재료명에 '소' 한 글자만 포함된 제품을 찾을 수 없습니다.")
    print("\n")


except FileNotFoundError as e:
    print(f"❌ 파일 오류: '{e.filename}' 파일을 찾을 수 없습니다.")
except Exception as e:
    print(f"오류가 발생했습니다: {e}")
```
