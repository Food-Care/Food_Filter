```python
# 원재료명 확인 코드
import json

def find_ingredients_with_keywords(data_path, keywords_to_find):
    """
    JSON 데이터 파일에서 특정 키워드를 포함하는 모든 원재료명을 찾아 출력합니다.

    :param data_path: 원본 식품 데이터 JSON 파일 경로
    :param keywords_to_find: 검색할 키워드 리스트
    """
    try:
        with open(data_path, "r", encoding="utf-8") as f:
            food_data = json.load(f)
        print(f"✅ '{data_path}' 파일을 성공적으로 불러왔습니다.")
    except FileNotFoundError:
        print(f"🚨 파일 오류: '{data_path}' 파일을 찾을 수 없습니다. 파일명을 확인해주세요.")
        return

    found_ingredients = set()

    for product in food_data:
        ingredients = product.get("원재료명", [])
        if not ingredients:
            continue

        for ingredient in ingredients:
            for keyword in keywords_to_find:
                if keyword in ingredient:
                    found_ingredients.add(ingredient)
                    break

    if found_ingredients:
        print(f"\n--- 🔍 총 {len(found_ingredients)}개의 원재료명을 찾았습니다 ---")
        for item in sorted(list(found_ingredients)):
            print(item)
    else:
        print("\n--- 해당 키워드를 포함하는 원재료를 찾지 못했습니다. ---")


if __name__ == "__main__":
    SEARCH_KEYWORDS = ["복숭"]

    print(f"지정한 키워드: {SEARCH_KEYWORDS}")
    
    input_file = "../Data_Storage/정제 데이터/식품데이터_최종_표준화_완료.json"

    find_ingredients_with_keywords(input_file, SEARCH_KEYWORDS)
```

```python
import json
import re

def load_json_data(file_path):
    """지정된 경로의 JSON 파일을 읽어 데이터를 반환합니다."""
    try:
        with open(file_path, 'r', encoding='utf-8') as f:
            return json.load(f)
    except FileNotFoundError:
        print(f"오류: '{file_path}' 파일을 찾을 수 없습니다.")
        return None
    except json.JSONDecodeError:
        print(f"오류: '{file_path}' 파일이 올바른 JSON 형식이 아닙니다.")
        return None

def count_unique_individual_ingredients(food_data):
    """
    전체 데이터에서 고유한 '개별 원재료'의 총 개수를 계산합니다.
    """
    if not food_data:
        return set()

    # 모든 개별 원재료를 중복 없이 저장할 set
    all_individual_ingredients = set()
    
    product_list = food_data

    for item in product_list:
        if isinstance(item, dict) and '원재료명' in item and item['원재료명']:
            ingredient_data = item['원재료명']
            
            # '원재료명'이 list 타입인 경우, 쉼표로 구분된 string으로 변환
            if isinstance(ingredient_data, list):
                ingredient_string = ', '.join(map(str, ingredient_data))
            else:
                ingredient_string = str(ingredient_data)

            # 쉼표(,)를 기준으로 원재료명을 분리
            individual_items = ingredient_string.split(',')

            # 분리된 각 개별 원재료를 set에 추가
            for ingredient in individual_items:
                # 앞뒤 공백 제거 및 빈 문자열 제외
                cleaned_ingredient = ingredient.strip()
                if cleaned_ingredient:
                    all_individual_ingredients.add(cleaned_ingredient)
    
    return all_individual_ingredients


# --- 메인 실행 부분 ---
if __name__ == "__main__":
    # 1. 파일 경로 설정
    food_data_path = '../Data_Storage/정제 데이터/식품데이터_카테고리_표준화.json'

    # 2. 데이터 로드
    food_data_json = load_json_data(food_data_path)

    if food_data_json:
        # 3. 고유한 개별 원재료 목록 추출
        unique_set = count_unique_individual_ingredients(food_data_json)
        
        # 4. 최종 결과 출력
        print(f"✅ 전체 식품 데이터에서 찾은 고유한 '개별 원재료'의 총개수는 {len(unique_set)}개입니다.")
        
        print("\n--- (예시) 고유 개별 원재료 목록 (알파벳, 가나다순 50개) ---")
        print(sorted(list(unique_set))[:50])
        print("----------------------------------------------------")
```

```python
import json

# --- 파일 로딩 및 개별 원재료 추출 함수 (이전과 동일) ---
def load_json_data(file_path):
    try:
        with open(file_path, 'r', encoding='utf-8') as f:
            return json.load(f)
    except FileNotFoundError:
        print(f"오류: '{file_path}' 파일을 찾을 수 없습니다.")
        return None
    except json.JSONDecodeError:
        print(f"오류: '{file_path}' 파일이 올바른 JSON 형식이 아닙니다.")
        return None

def get_unique_individual_ingredients(food_data):
    if not food_data: return set()
    all_ingredients = set()
    product_list = food_data
    for item in product_list:
        if isinstance(item, dict) and '원재료명' in item and item['원재료명']:
            ingredient_data = item['원재료명']
            if isinstance(ingredient_data, list):
                ingredient_string = ', '.join(map(str, ingredient_data))
            else:
                ingredient_string = str(ingredient_data)
            individual_items = ingredient_string.split(',')
            for ingredient in individual_items:
                cleaned_ingredient = ingredient.strip()
                if cleaned_ingredient:
                    all_ingredients.add(cleaned_ingredient)
    return all_ingredients

# --- 새로운 최종 분석 및 결과 출력 함수 ---
def analyze_and_report_individuals(ingredients_set, allergy_db):
    """
    고유 개별 원재료 목록을 알레르기 규칙에 따라 세 그룹으로 분류하고 결과를 출력합니다.
    """
    print("\n--- 🔬 15,243개 개별 원재료 대상, 최종 규칙 분석 시작 ---")
    
    matched_by_keyword = set()
    matched_by_exclusion = set()

    # 1. 키워드 및 예외 키워드에 걸리는 항목들을 각각 찾기
    for ingredient in ingredients_set:
        # 키워드 검사
        for allergen, data in allergy_db.items():
            if "keyword" in data:
                for keyword in data["keyword"]:
                    if keyword in ingredient:
                        matched_by_keyword.add(ingredient)
                        break
        
        # 예외 키워드 검사
        for allergen, data in allergy_db.items():
            if "exclusions" in data:
                for exclusion in data["exclusions"]:
                    if exclusion in ingredient:
                        matched_by_exclusion.add(ingredient)
                        break

    # 2. 어디에도 걸리지 않은 원재료 계산
    # 전체 집합에서 키워드 일치 집합과 예외 일치 집합을 모두 뺌
    unmatched_ingredients = ingredients_set.difference(matched_by_keyword).difference(matched_by_exclusion)
    
    # 3. 최종 결과 요약 출력
    print("\n--- 📊 최종 분석 결과 요약 ---")
    print(f"1. 분석 대상 고유 개별 원재료: {len(ingredients_set)}개")
    print(f"2. ✅ '키워드' 일치: {len(matched_by_keyword)}개")
    print(f"3. ⚠️ '예외 키워드' 일치: {len(matched_by_exclusion)}개")
    print(f"4. ❌ '규칙 미적용': {len(unmatched_ingredients)}개")
    print("---------------------------------\n")

    # 4. '규칙 미적용' 데이터 상세 목록 출력 (가장 중요한 부분)
    print("--- ❌ '규칙 미적용' 개별 원재료 목록 (예시 200개) ---")
    print("   (이 목록을 검토하여 새로운 알레르기 키워드를 추가하거나 규칙을 보완할 수 있습니다.)")
    print(sorted(list(unmatched_ingredients))[:200])
    print("\n----------------------------------------------------")


# --- 메인 실행 부분 ---
if __name__ == "__main__":
    # 1. 파일 경로 설정
    allergy_rules_path = '../Data_Storage/표준화 데이터/Allergy_ver_2.json'
    food_data_path = '../Data_Storage/정제 데이터/식품데이터_카테고리_표준화.json'

    # 2. 데이터 로드
    allergy_data = load_json_data(allergy_rules_path)
    food_data_json = load_json_data(food_data_path)

    if allergy_data and food_data_json:
        # 3. 고유한 개별 원재료 목록 (15,243개) 준비
        unique_individual_set = get_unique_individual_ingredients(food_data_json)
        print(f"✅ 분석 준비 완료: {len(unique_individual_set)}개의 고유 개별 원재료를 찾았습니다.")
        
        # 4. 최종 분석 및 결과 리포트 실행
        analyze_and_report_individuals(unique_individual_set, allergy_data)
```

```python
import json

def load_json_data(file_path):
    """지정된 경로의 JSON 파일을 읽어 데이터를 반환합니다."""
    try:
        with open(file_path, 'r', encoding='utf-8') as f:
            return json.load(f)
    except FileNotFoundError:
        print(f"🚨 오류: '{file_path}' 파일을 찾을 수 없습니다.")
        return None
    except json.JSONDecodeError:
        print(f"🚨 오류: '{file_path}' 파일이 올바른 JSON 형식이 아닙니다.")
        return None

def get_unique_individual_ingredients(food_data):
    """전체 데이터에서 고유한 '개별 원재료'의 set을 반환합니다."""
    if not food_data: return set()
    all_ingredients = set()
    for item in food_data:
        if isinstance(item, dict) and '원재료명' in item and item['원재료명']:
            ingredient_data = item['원재료명']
            if isinstance(ingredient_data, list):
                ingredient_string = ', '.join(map(str, ingredient_data))
            else:
                ingredient_string = str(ingredient_data)
            individual_items = ingredient_string.split(',')
            for ingredient in individual_items:
                cleaned_ingredient = ingredient.strip()
                if cleaned_ingredient:
                    all_ingredients.add(cleaned_ingredient)
    return all_ingredients

def find_unmatched_ingredients(ingredients_set, allergy_db):
    """
    알레르기 규칙(키워드, 예외)에 해당하지 않는 원재료만 필터링하여 반환합니다.
    """
    # 키워드나 예외 키워드에 하나라도 걸리는 원재료를 저장할 set
    filtered_out_ingredients = set()

    for ingredient in ingredients_set:
        is_filtered = False
        for allergen, data in allergy_db.items():
            # 키워드 검사
            if "keyword" in data:
                for keyword in data["keyword"]:
                    if keyword in ingredient:
                        filtered_out_ingredients.add(ingredient)
                        is_filtered = True
                        break
            if is_filtered: break

            # 예외 키워드 검사 (키워드에 걸리지 않았을 경우에만)
            if "exclusions" in data:
                for exclusion in data["exclusions"]:
                    if exclusion in ingredient:
                        filtered_out_ingredients.add(ingredient)
                        is_filtered = True
                        break
            if is_filtered: break

    # 전체 집합에서 걸러진 것들을 제외하여 최종 목록 생성
    unmatched_set = ingredients_set.difference(filtered_out_ingredients)
    return unmatched_set


# --- 메인 실행 부분 ---
if __name__ == "__main__":
    # 1. 파일 경로 설정
    allergy_rules_path = '../Data_Storage/표준화 데이터/Allergy_ver_2.json'
    food_data_path = '../Data_Storage/정제 데이터/식품데이터_최종_표준화_완료.json'

    # 2. 데이터 로드
    allergy_data = load_json_data(allergy_rules_path)
    food_data_json = load_json_data(food_data_path)

    if allergy_data and food_data_json:
        # 3. 고유한 개별 원재료 목록 준비
        unique_ingredients = get_unique_individual_ingredients(food_data_json)
        print(f"✅ 분석 준비 완료: 총 {len(unique_ingredients)}개의 고유 개별 원재료를 찾았습니다.")

        # 4. 규칙에 걸러지지 않은 원재료 필터링
        unmatched_list = find_unmatched_ingredients(unique_ingredients, allergy_data)

        # 5. 최종 결과 출력
        print("\n" + "="*50)
        print(f"📊 최종 분석 결과")
        print("="*50)
        print(f"키워드 규칙에 걸러지지 않은 고유 원재료는 총 {len(unmatched_list)}개입니다.")
        print("아래는 해당 원재료의 전체 목록입니다.")
        print("-"*50 + "\n")

        # 정렬하여 전체 목록 출력
        for ingredient_item in sorted(list(unmatched_list)):
            print(ingredient_item)
```
